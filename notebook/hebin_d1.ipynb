# DATA CLEANING

Count missing value per column

```python
df.isnull().sum()
```

```python
df.isnull().sum().sum()
```

Express missingness as a proportion

```python
df.isnull().mean()
```

Inspect the affected rows

```python
df[df['physical_activity'].isnull()]
```

```python
df[df.isnull().any(axis=1)]
```

Count the placeholder

```python
(df['glucose'] == 0).sum()
```

```python
placeholder_cols = ['glucose','blood_pressure','skin_thickness','insulin','bmi']
for c in placeholder_cols:
    print(c,'->',(df[c] == 0).sum(), 'zeros')
```

Convert placeholders to missing values

```python
import numpy as np
```

```python
df['glucose'] = df['glucose'].replace(0, np.nan)
```

```python
before_missing = df.isnull().sum().sum()

for c in placeholder_cols:
    df[c]=df[c].replace(0, np.nan)

after_missing = df.isnull().sum().sum()
print('missing cells before:', before_missing)
print('missing cells after:', after_missing)
```

Re-measure the damage

```python
df.isnull().sum()
```

```python
df.isnull().mean()
```

Fill numeric columns

```python
df['glucose']=df['glucose'].fillna(df['glucose'].mean())
```

```python
column_fill=['age','blood_pressure','skin_thickness','bmi','insulin']

for c in column_fill:
    df[c]=df[c].fillna(df[c].median())
```

```python
df.isnull().sum()
```

Fill categorical columns

```python
categorical_fill=['physical_activity','smoker']

for c in categorical_fill:
    df[c]=df[c].fillna('unknown')
```

```python
df.isnull().sum()
```

Verify no missing values remain

```python
df.isnull().sum().sum()
```

```python
df.isnull().sum()
```

```python
df['gender'].unique()
```

Remove stray whitespace

```python
df['gender']=df['gender'].str.strip()
```

```python
df['gender'].unique()
```

```python
df.value_counts('gender')
```

```python
categorical_fill=['physical_activity','smoker']

for c in categorical_fill:
    df[c]=df[c].str.strip()
```

```python
df.value_counts("smoker")
```

Standardise capitalisation

```python
df['gender']=df['gender'].str.lower()
```

```python
df.value_counts("gender")
```

```python

df['physical_activity'].unique()
```

```python
df['physical_activity']=df['physical_activity'].str.title()
```

```python
df.value_counts("physical_activity")
```

```python
df['smoker']=df['smoker'].str.upper()
```

```python
df.value_counts('smoker')
```

Unify equivalent categories

```python
df['gender']=df['gender'].replace({'f':'female','m':'male'})
```

```python
df.value_counts('gender')
```

```python
df['smoker']=df['smoker'].replace({'N':'NO','Y':'YES'})
```

```python
df.value_counts('smoker')
```

Convert text dates to real dates

```python
df.value_counts('last_checkup_date')
```

```python
df['last_checkup_date']=pd.to_datetime(df['last_checkup_date'],dayfirst=True)
```

```python
df.value_counts('last_checkup_date')
```

Correct remaining type mismatches
